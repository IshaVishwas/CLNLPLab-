# Experiment-2: Basic Text Preprocessing

### Experiment 2.1: Basic Text Preprocessing

```python
import string

def experiment_2_1(file_path):
    with open(file_path, 'r') as f:
        text = f.read()

    upper_count = sum(1 for c in text if c.isupper())
    text = text.lower()

    puncs = [c for c in text if c in string.punctuation]
    punc_types = set(puncs)
    text = "".join(c for c in text if c not in string.punctuation)

    digits = [c for c in text if c.isdigit()]
    text = "".join(c for c in text if not c.isdigit())

    cleaned = " ".join(text.split())
    ws_removed = len(text) - len(cleaned)

    print("--- Experiment 2.1 Full Results ---")
    print(f"Uppercase: {upper_count}, Punc Types: {len(punc_types)}, Digits: {len(digits)}, Whitespace: {ws_removed}")
    print(f"Full Cleaned Text:\n{cleaned}")

path_2_1 = "2.1_text_data.txt"
experiment_2_1(path_2_1)
```

### Experiment 2.2: Tokenisation

```python
import nltk
import spacy
import re

# Download required NLTK tokenizer
nltk.download('punkt_tab', quiet=True)

# Load spaCy English model
nlp = spacy.load('en_core_web_sm')

# Read the text file
with open("2.2_tokenization_data.txt", 'r') as f:
    raw_text = f.read()

print("--- Experiment 2.2 Full Tokenization ---")

# NLTK
print("\n[NLTK Word Tokens]:")
print(nltk.word_tokenize(raw_text))

print("\n[NLTK Sentence Tokens]:")
print(nltk.sent_tokenize(raw_text))

# spaCy
doc = nlp(raw_text)

print("\n[spaCy Word Tokens]:")
print([t.text for t in doc])

print("\n[spaCy Sentence Tokens]:")
print([s.text for s in doc.sents])

# Pure Python
words = raw_text.split()
sents = re.split(r'(?<=[.!?]) +', raw_text)

print("\n[Python Word Tokens]:")
print(words)

print("\n[Python Sentence Tokens]:")
print(sents)
```

### Experiment 2.3: Stop Words Removal

```python
import nltk
from nltk.corpus import stopwords

# Download stop words
nltk.download('stopwords', quiet=True)

# Read the text file
with open("2.3_clean_data.txt", 'r') as f:
    text = f.read().lower()

# Tokenize the text
tokens = nltk.word_tokenize(text)

# Get English stop words
stop_words = set(stopwords.words('english'))

# Find stop words present in the text
isolated_stops = [t for t in tokens if t in stop_words]

# Remove stop words and punctuation
filtered = [
    t for t in tokens
    if t not in stop_words and t.isalnum()
]

print("--- Experiment 2.3 Full Results ---")

print(f"Original Tokens:\n{tokens}")

print(f"\nIsolated Stop Words Found:\n{list(set(isolated_stops))}")

print(f"\nFiltered Tokens (No Stop Words):\n{filtered}")
```
