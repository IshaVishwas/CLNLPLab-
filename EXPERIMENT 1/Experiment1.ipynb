# Experiment 1: Data Analysis

```python
file_path = 'employee_information_100.csv'
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

file_path = 'employee_information_100.csv'

df = pd.read_csv(file_path)

print("First 5 rows:")
display(df.head())

print("\nDataset Information:")
df.info()
```

# Experiment 1.1: Average Salary by Department

```python
avg_salary = df.groupby('Department')['Salary'].mean().sort_values()

print("Average salary by department:")
print(avg_salary)

plt.figure(figsize=(10, 6))
avg_salary.plot(kind='barh')

plt.title('Average Salary by Department')
plt.xlabel('Average Salary')
plt.ylabel('Department')

plt.show()
```

# Experiment 1.2: Number of Employees per Department

```python
dept_counts = df['Department'].value_counts()

print("Employee counts per department:")
print(dept_counts)

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Department')

plt.title('Number of Employees in Each Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')

plt.show()
```

# Experiment 1.3: Gender Distribution

```python
gender_counts = df['Gender'].value_counts()

print("Gender distribution:")
print(gender_counts)

plt.figure(figsize=(7, 7))

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%',
    colors=['lightblue', 'pink']
)

plt.title('Gender Distribution')

plt.show()
```

# Experiment 1.4: Salary Distribution Histogram

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df['Salary'],
    bins=15,
    kde=True
)

plt.title('Distribution of Employee Salaries')
plt.xlabel('Salary')
plt.ylabel('Frequency')

plt.show()
```

# Experiment 1.5: Relationship between Experience and Salary

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Experience_Years',
    y='Salary',
    hue='Department',
    style='Gender'
)

plt.title('Experience vs Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')

plt.legend(loc='upper left')

plt.show()
```

# Experiment 1.6: Top 10 Highest Paid Employees

```python
top_10_salaries = df.nlargest(
    10,
    'Salary'
)[['Name', 'Department', 'Salary', 'Experience_Years']]

print("Top 10 employees with the highest salary:")

display(top_10_salaries)
```

# Experiment 1.7: Highest Salary in Every Department

```python
max_salary_dept = df.groupby('Department')['Salary'].max()

print("Highest salary in each department:")
print(max_salary_dept)
```

# Experiment 1.8: Employees with Above-Average Salary

```python
overall_avg = df['Salary'].mean()

high_earners = df[df['Salary'] > overall_avg]

print("Overall average salary:", overall_avg)

print(
    "Number of employees earning above average:",
    len(high_earners)
)

display(high_earners.head())
```

# Experiment 1.9: Average Experience per Department

```python
avg_exp = df.groupby('Department')['Experience_Years'].mean()

print("Average years of experience by department:")
print(avg_exp)
```

# Experiment 1.10: Age Distribution Histogram

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df['Age'],
    bins=15,
    kde=True
)

plt.title('Distribution of Employee Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.show()
```
